# 1. Business Understanding

## About Sapphire

SAPPHIRE is a Pakistani fashion and lifestyle retail brand that offers products across apparel and lifestyle categories, including women's wear, menswear, kidswear, western wear, home textiles, accessories, beauty and fragrances. The brand operates through physical retail stores as well as an online e-store.

SAPPHIRE was launched as a retail brand in 2014, building on the wider Sapphire Group's textile-industry background.

## Business Model

SAPPHIRE primarily follows a **B2C (Business-to-Consumer) retail model**, where products are sold directly to individual customers through retail stores and online channels.

The basic business process can be represented as:

**Products → Customers → Orders → Sales Revenue**

The business needs to understand product demand, sales performance, pricing, discounts, customer ordering patterns, locations and purchasing behavior in order to support better business decisions.

3.1. How did total revenue change year over year?

```python
ys = sales[
    (sales['Date'] >= '2024-01-01') &
    (sales['Date'] < '2026-10-01')
]

yearly_sales = ys.groupby(
    ys['Date'].dt.year
)['Revenue_PKR'].sum()

q1 = pd.DataFrame({
    'Year': yearly_sales.index,
    'Total Revenue (PKR)': yearly_sales.values,
    'YoY Change (PKR)': yearly_sales.diff().values,
    'YoY Change (%)': yearly_sales.pct_change().mul(100).values
})

q1['YoY Change (PKR)'] = q1['YoY Change (PKR)'].round(2)
q1['Total Revenue (PKR)'] = q1['Total Revenue (PKR)'].round(2)
q1['YoY Change (%)'] = q1['YoY Change (%)'].round(2)

print(q1)
```

3.2 Which months generated the highest and lowest revenue during the historical sales period from January 2024 to Sep 2026?

```python
monthly_sales = sales[
    (sales['Date'] >= '2024-01-01') &
    (sales['Date'] < '2026-10-01')
].groupby(sales['Date'].dt.month_name())['Revenue_PKR'].sum()

monthly_sales = monthly_sales.sort_values(ascending=False)

maxsales = monthly_sales.max()
minsales = monthly_sales.min()

maxmonth = monthly_sales.idxmax()
minmonth = monthly_sales.idxmin()

print('Results:')
print('Highest Revenue:', maxsales)
print('Highest Revenue Month:', maxmonth)
print('Lowest Revenue:', minsales)
print('Lowest Revenue Month:', minmonth)
```

3.3 Which product categories generated the highest total revenue and the highest number of units sold during the historical sales period?

```python
salesper = sales.groupby('Category')[
    ['Revenue_PKR', 'Units_Sold']
].sum()

salesperrevenuemax = salesper['Revenue_PKR'].idxmax()
salesperunitsmax = salesper['Units_Sold'].idxmax()

actualrev = salesper['Revenue_PKR'].max()
actualunits = salesper['Units_Sold'].max()

print('Results:')
print('Highest Revenue:', actualrev)
print('Category with Highest Revenue:', salesperrevenuemax)
print('Highest Units Sold:', actualunits)
print('Category with Highest Units Sold:', salesperunitsmax)
```

3.4 Which 10 SKUs generated the highest total revenue and which 10 SKUs recorded the highest unit sales during the historical sales period?

3.5 Which product categories showed the strongest seasonal changes in revenue during the historical sales period?

```python
seasonal_sales = sales[
    (sales['Date'] >= '2024-01-01') &
    (sales['Date'] < '2026-10-01')
].copy()

seasonal_sales['Year_Month'] = seasonal_sales['Date'].dt.to_period('M')
seasonal_sales['Month'] = seasonal_sales['Date'].dt.month_name()


monthly_category = seasonal_sales.groupby(
    ['Category', 'Year_Month', 'Month']
)['Revenue_PKR'].sum().reset_index()

avg_monthly_revenue = monthly_category.groupby(
    ['Category', 'Month']
)['Revenue_PKR'].mean().reset_index()


seasonal_range = avg_monthly_revenue.groupby('Category').agg(
    Highest_Monthly_Revenue=('Revenue_PKR', 'max'),
    Lowest_Monthly_Revenue=('Revenue_PKR', 'min')
)

seasonal_range['Seasonal_Revenue_Range'] = (
    seasonal_range['Highest_Monthly_Revenue'] -
    seasonal_range['Lowest_Monthly_Revenue']
)

seasonal_range = seasonal_range.sort_values(
    by='Seasonal_Revenue_Range',
    ascending=False
).round(2)

print('Q5: Seasonal Revenue Variation by Category')
display(seasonal_range)

print('Category with Strongest Seasonal Variation:',
      seasonal_range['Seasonal_Revenue_Range'].idxmax())
```

3.9 What relationship existed between base price and units sold across SKUs during the historical sales period?

```python
salesof = sales[['SKU', 'Base_Price_PKR', 'Units_Sold']]

salespersku = sales.groupby('SKU').agg({
    'Base_Price_PKR': 'mean',
    'Units_Sold': 'sum'
})

salespersku = salespersku.sort_values(
    by='Base_Price_PKR',
    ascending=False
)

priceunitscorr = salespersku['Base_Price_PKR'].corr(
    salespersku['Units_Sold']
)

print('Q9: Relationship Between Base Price and Units Sold')
print('Results:')

print('SKU-wise Base Price and Units Sold:')
display(salespersku)

print('Correlation between Base Price and Units Sold:',
      round(priceunitscorr, 3))

if priceunitscorr > 0:
    print('Result: A positive relationship exists between base price and units sold.')
elif priceunitscorr < 0:
    print('Result: A negative relationship exists between base price and units sold.')
else:
    print('Result: No linear relationship exists between base price and units sold.')
```

**Result:** The correlation coefficient between base price and units sold is 0.061, indicating a very weak positive relationship. This suggests that base price has almost no linear association with the number of units sold across SKUs. Therefore, base price alone may not be a useful indicator of sales volume.

```python
discount_units = sales.groupby('Discount')['Units_Sold'].mean()

discount_units_corr = sales['Discount'].corr(sales['Units_Sold'])

print('Q10: Relationship Between Discount and Units Sold')
print('Results:')
print('Average Units Sold at Each Discount Level:')
display(discount_units.to_frame(name='Average_Units_Sold'))

print('Correlation between Discount and Units Sold:',
      round(discount_units_corr, 3))

if discount_units_corr > 0:
    print('Result: A positive relationship exists between discount and units sold.')
elif discount_units_corr < 0:
    print('Result: A negative relationship exists between discount and units sold.')
else:
    print('Result: No linear relationship exists between discount and units sold.')
```

3.13 Which base-price band generated the highest revenue and unit sales during the historical sales period?

```python
sku_price = sales.groupby('SKU').agg({
    'Base_Price_PKR': 'mean',
    'Revenue_PKR': 'sum',
    'Units_Sold': 'sum'
})

sku_price['Price_Band'] = pd.qcut(
    sku_price['Base_Price_PKR'].rank(method='first'),
    q=3,
    labels=['Low', 'Medium', 'High']
)

price_band_analysis = sku_price.groupby(
    'Price_Band', observed=False
).agg({
    'Revenue_PKR': 'sum',
    'Units_Sold': 'sum'
}).rename(columns={
    'Revenue_PKR': 'Total_Revenue',
    'Units_Sold': 'Total_Units_Sold'
}).round(2)

print('Q13: Revenue and Units Sold by Base-Price Band')
print('Results:')
display(price_band_analysis)

print('Highest Revenue Price Band:',
      price_band_analysis['Total_Revenue'].idxmax())

print('Highest Units Sold Price Band:',
      price_band_analysis['Total_Units_Sold'].idxmax())
```

3.14 How many SKUs can be classified as high-, medium-, and low-performing based on their total unit sales during the historical sales period?

```python
sku_performance = sales.groupby('SKU')['Units_Sold'].sum().to_frame()

sku_performance['Performance'] = pd.qcut(
    sku_performance['Units_Sold'].rank(method='first'),
    q=3,
    labels=['Low', 'Medium', 'High']
)

performance_count = sku_performance['Performance'].value_counts().reindex(
    ['High', 'Medium', 'Low'], fill_value=0
)

print('Q14: SKU Performance Classification')
print('Results:')

print('Number of SKUs in Each Performance Group:')
display(performance_count.to_frame(name='Number_of_SKUs'))

print('Total SKUs:', len(sku_performance))
print('High-Performing SKUs:', performance_count['High'])
print('Medium-Performing SKUs:', performance_count['Medium'])
print('Low-Performing SKUs:', performance_count['Low'])
```

3.15 Which months showed unusually high or unusually low revenue compared with the overall monthly revenue pattern from January 2024 to Sep 2026?


```python
monthly_revenue = sales[
    (sales['Date'] >= '2024-01-01') &
    (sales['Date'] < '2026-10-01')
].groupby(
    sales.loc[
        (sales['Date'] >= '2024-01-01') &
        (sales['Date'] < '2026-10-01'),
        'Date'
    ].dt.to_period('M')
)['Revenue_PKR'].sum()

monthly_results = monthly_revenue.to_frame(name='Monthly_Revenue')

overall_mean = monthly_revenue.mean()
overall_std = monthly_revenue.std()

high_limit = overall_mean + (2 * overall_std)
low_limit = overall_mean - (2 * overall_std)

monthly_results['Status'] = 'Normal'

monthly_results.loc[
    monthly_results['Monthly_Revenue'] > high_limit, 'Status'
] = 'Unusually High'

monthly_results.loc[
    monthly_results['Monthly_Revenue'] < low_limit, 'Status'
] = 'Unusually Low'

print('Q15: Unusual Monthly Revenue Analysis')
print('Results:')
display(monthly_results.round(2))

unusual_months = monthly_results[
    monthly_results['Status'] != 'Normal'
]

print('Unusually High or Low Revenue Months:')
display(unusual_months.round(2))

print('Total Unusual Months:', len(unusual_months))
```

3.16 Which SKUs recorded unusually high unit sales or revenue compared with other SKUs in the same category during the historical sales period?

```python
# Q16: Unusual SKU Performance Within Each Category

# Step 1: Calculate total units and revenue for each SKU
category_sku = sales.groupby(
    ['Category', 'SKU'],
    observed=True
).agg({
    'Units_Sold': 'sum',
    'Revenue_PKR': 'sum'
}).reset_index()

# Step 2: Initialize unusual flags
category_sku['Unusual_Units'] = False
category_sku['Unusual_Revenue'] = False

# Step 3: Check each category separately
for category in category_sku['Category'].unique():

    category_index = category_sku.index[
        category_sku['Category'] == category
    ]

    category_data = category_sku.loc[category_index]

    if len(category_data) < 4:
        continue

    # Units Sold: IQR
    q1_units = category_data['Units_Sold'].quantile(0.25)
    q3_units = category_data['Units_Sold'].quantile(0.75)
    iqr_units = q3_units - q1_units

    lower_units = q1_units - 1.5 * iqr_units
    upper_units = q3_units + 1.5 * iqr_units

    category_sku.loc[category_index, 'Unusual_Units'] = (
        (category_data['Units_Sold'] < lower_units) |
        (category_data['Units_Sold'] > upper_units)
    ).values

    # Revenue: IQR
    q1_revenue = category_data['Revenue_PKR'].quantile(0.25)
    q3_revenue = category_data['Revenue_PKR'].quantile(0.75)
    iqr_revenue = q3_revenue - q1_revenue

    lower_revenue = q1_revenue - 1.5 * iqr_revenue
    upper_revenue = q3_revenue + 1.5 * iqr_revenue

    category_sku.loc[category_index, 'Unusual_Revenue'] = (
        (category_data['Revenue_PKR'] < lower_revenue) |
        (category_data['Revenue_PKR'] > upper_revenue)
    ).values

# Step 4: Select unusual SKUs
unusual_skus = category_sku[
    category_sku['Unusual_Units'] |
    category_sku['Unusual_Revenue']
].copy()

# Step 5: Explain why each SKU was flagged
unusual_skus['Unusual_Reason'] = unusual_skus.apply(
    lambda row:
        'Units Sold and Revenue'
        if row['Unusual_Units'] and row['Unusual_Revenue']
        else 'Units Sold'
        if row['Unusual_Units']
        else 'Revenue',
    axis=1
)

# Step 6: Display results
print('Q16: Unusual SKU Performance Within Each Category')
print('Results:')

print('Total Unique SKUs:', category_sku['SKU'].nunique())
print('SKUs with Unusual Units Sold:',
      int(category_sku['Unusual_Units'].sum()))
print('SKUs with Unusual Revenue:',
      int(category_sku['Unusual_Revenue'].sum()))
print('Total SKUs Flagged as Unusual:', len(unusual_skus))

print('\nCategory-wise SKU Counts:')
display(
    category_sku.groupby('Category', observed=True)['SKU']
    .nunique()
    .to_frame(name='Unique_SKUs')
)

print('\nUnusual SKU Details:')
display(
    unusual_skus[
        [
            'Category',
            'SKU',
            'Units_Sold',
            'Revenue_PKR',
            'Unusual_Units',
            'Unusual_Revenue',
            'Unusual_Reason'
        ]
    ].sort_values(
        by=['Category', 'Revenue_PKR'],
        ascending=[True, False]
    ).reset_index(drop=True)
)

print('\nFinal Result:')
if len(unusual_skus) == 0:
    print('No unusual SKU performance was detected using the IQR method.')
else:
    print(
        len(unusual_skus),
        'SKUs were flagged for unusual units sold or revenue '
        'compared with other SKUs in their category.'
    )
```

```python

# Visualization 15: Unusual Monthly Revenue

# Step 1: Filter historical period
monthly_data = sales[
    (sales['Date'] >= '2024-01-01') &
    (sales['Date'] < '2026-10-01')
].copy()

# Step 2: Calculate total revenue per month
monthly_data['Year_Month'] = monthly_data['Date'].dt.to_period('M')

monthly_revenue = monthly_data.groupby(
    'Year_Month'
)['Revenue_PKR'].sum().reset_index()

monthly_revenue['Year_Month'] = (
    monthly_revenue['Year_Month'].astype(str)
)

# Step 3: Calculate unusual revenue limits
average_revenue = monthly_revenue['Revenue_PKR'].mean()
std_revenue = monthly_revenue['Revenue_PKR'].std()

upper_limit = average_revenue + (2 * std_revenue)
lower_limit = average_revenue - (2 * std_revenue)

monthly_revenue['Status'] = 'Normal'

monthly_revenue.loc[
    monthly_revenue['Revenue_PKR'] > upper_limit,
    'Status'
] = 'Unusually High'

monthly_revenue.loc[
    monthly_revenue['Revenue_PKR'] < lower_limit,
    'Status'
] = 'Unusually Low'

# Step 4: Create visualization
plt.figure(figsize=(15, 7))

sns.lineplot(
    data=monthly_revenue,
    x='Year_Month',
    y='Revenue_PKR',
    marker='o',
    label='Monthly Revenue'
)

# Highlight unusual months
high_months = monthly_revenue[
    monthly_revenue['Status'] == 'Unusually High'
]

low_months = monthly_revenue[
    monthly_revenue['Status'] == 'Unusually Low'
]

plt.scatter(
    high_months['Year_Month'],
    high_months['Revenue_PKR'],
    s=120,
    marker='^',
    label='Unusually High'
)

plt.scatter(
    low_months['Year_Month'],
    low_months['Revenue_PKR'],
    s=120,
    marker='v',
    label='Unusually Low'
)

plt.axhline(
    average_revenue,
    linestyle='--',
    label='Average Revenue'
)

plt.title('SAPPHIRE Monthly Revenue: Unusual Patterns (2024–Sep 2026)',
          fontsize=15)
plt.xlabel('Month')
plt.ylabel('Revenue (PKR)')
plt.xticks(rotation=60)
plt.ticklabel_format(axis='y', style='plain')
plt.legend()
plt.tight_layout()
plt.show()

# Step 5: Display results
print('Q15: Unusual Monthly Revenue Analysis')
print('Results:')

display(monthly_revenue.round(2))

unusual_months = monthly_revenue[
    monthly_revenue['Status'] != 'Normal'
]

print('Number of Unusual Months:', len(unusual_months))
print('Unusual Months:')
display(unusual_months.round(2))
```

**Category-Wise Identification of Unusual SKU Performance**

```python

# Visualization 16: Unusual SKU Performance by Category

# Step 1: Calculate SKU-level performance
category_sku = sales.groupby(
    ['Category', 'SKU'],
    observed=True
).agg({
    'Units_Sold': 'sum',
    'Revenue_PKR': 'sum'
}).reset_index()

# Step 2: Identify unusual SKUs using category-wise IQR
category_sku['Unusual_Units'] = False
category_sku['Unusual_Revenue'] = False

for category in category_sku['Category'].dropna().unique():

    category_index = category_sku.index[
        category_sku['Category'] == category
    ]

    category_data = category_sku.loc[category_index]

    if len(category_data) < 4:
        continue

    # Units Sold limits
    q1_units = category_data['Units_Sold'].quantile(0.25)
    q3_units = category_data['Units_Sold'].quantile(0.75)
    iqr_units = q3_units - q1_units

    lower_units = q1_units - 1.5 * iqr_units
    upper_units = q3_units + 1.5 * iqr_units

    category_sku.loc[category_index, 'Unusual_Units'] = (
        (category_data['Units_Sold'] < lower_units) |
        (category_data['Units_Sold'] > upper_units)
    ).values

    # Revenue limits
    q1_revenue = category_data['Revenue_PKR'].quantile(0.25)
    q3_revenue = category_data['Revenue_PKR'].quantile(0.75)
    iqr_revenue = q3_revenue - q1_revenue

    lower_revenue = q1_revenue - 1.5 * iqr_revenue
    upper_revenue = q3_revenue + 1.5 * iqr_revenue

    category_sku.loc[category_index, 'Unusual_Revenue'] = (
        (category_data['Revenue_PKR'] < lower_revenue) |
        (category_data['Revenue_PKR'] > upper_revenue)
    ).values

# Step 3: Create a combined status
category_sku['Unusual_Status'] = 'Normal'

category_sku.loc[
    category_sku['Unusual_Units'] &
    category_sku['Unusual_Revenue'],
    'Unusual_Status'
] = 'Both'

category_sku.loc[
    category_sku['Unusual_Units'] &
    ~category_sku['Unusual_Revenue'],
    'Unusual_Status'
] = 'Units Only'

category_sku.loc[
    ~category_sku['Unusual_Units'] &
    category_sku['Unusual_Revenue'],
    'Unusual_Status'
] = 'Revenue Only'

# Step 4: Count unusual SKUs by category
category_summary = category_sku.groupby(
    'Category', observed=True
).agg(
    Total_SKUs=('SKU', 'nunique'),
    Unusual_Units=('Unusual_Units', 'sum'),
    Unusual_Revenue=('Unusual_Revenue', 'sum')
).reset_index()

print('Q16: Unusual SKU Performance by Category')
print('Results:')
display(category_summary)

# Step 5: Plot unusual SKU counts
category_summary.plot(
    x='Category',
    y=['Unusual_Units', 'Unusual_Revenue'],
    kind='bar',
    figsize=(13, 6)
)

plt.title('Unusual SKU Counts by Category', fontsize=16)
plt.xlabel('Product Category')
plt.ylabel('Number of Unusual SKUs')
plt.xticks(rotation=35, ha='right')
plt.legend(['Unusual Units Sold', 'Unusual Revenue'])
plt.tight_layout()
plt.show()

# Step 6: Show unusual SKU details
unusual_skus = category_sku[
    category_sku['Unusual_Units'] |
    category_sku['Unusual_Revenue']
].copy()

print('Total Unique SKUs:', category_sku['SKU'].nunique())
print('SKUs with Unusual Units Sold:',
      int(category_sku['Unusual_Units'].sum()))
print('SKUs with Unusual Revenue:',
      int(category_sku['Unusual_Revenue'].sum()))
print('Total SKUs Flagged as Unusual:', len(unusual_skus))

print('Unusual SKU Details:')
display(
    unusual_skus[
        [
            'Category',
            'SKU',
            'Units_Sold',
            'Revenue_PKR',
            'Unusual_Status'
        ]
    ].sort_values(
        by=['Category', 'Revenue_PKR'],
        ascending=[True, False]
    ).reset_index(drop=True)
)
```
